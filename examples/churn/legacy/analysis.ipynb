```python
import pickle

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

PATH = 'data/raw/churn.csv'   # hardcoded, works only from this folder
```

```python
# load
df = pd.read_csv(PATH)
df = df.dropna()
print(df.shape)
print(df.head())
```

```python
# quick stats - I needed these twice so they are here twice
cr = df['churn'].mean()
print('churn rate', cr)
print('overall churn rate is', df[df.churn == 1].shape[0] / df.shape[0])
print(df['contract_type'].value_counts())
```

```python
# encode by hand
contract_map = {'monthly': 0, 'one_year': 1, 'two_year': 2}
df['contract_num'] = df['contract_type'].map(contract_map)
features = ['tenure_months', 'monthly_charges', 'support_tickets', 'contract_num']
X = df[features]
y = df['churn']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=7)
```

```python
# the model that gave 91% in the meeting
model = RandomForestClassifier(n_estimators=500, random_state=7)
model.fit(X_train, y_train)
pred = model.predict(X_test)
print('accuracy', accuracy_score(y_test, pred))
```

```python
# save it
pickle.dump(model, open('model.pkl', 'wb'))
print('saved model.pkl')
```
